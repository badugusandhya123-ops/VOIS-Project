```python
# ============================================================
# EDA — LOAD CLEANED DATASET
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load cleaned dataset
input_path = "final_dataset.csv"

df = pd.read_csv(input_path)

print("=" * 70)
print("CLEANED DATASET LOADED")
print("=" * 70)

print(f"Rows    : {df.shape[0]:,}")
print(f"Columns : {df.shape[1]:,}")

display(df.head())
# ============================================================
# EDA PART 1 — DATASET OVERVIEW & DATA QUALITY
# ============================================================

print("=" * 70)
print("EDA PART 1 — DATASET OVERVIEW")
print("=" * 70)


# 1. Shape
print("\n1. DATASET SIZE")
print("-" * 70)

print(f"Rows    : {df.shape[0]:,}")
print(f"Columns : {df.shape[1]:,}")


# 2. Data types
print("\n2. DATA TYPES")
print("-" * 70)

display(
    pd.DataFrame({
        "Column": df.columns,
        "Data_Type": df.dtypes.astype(str)
    })
)


# 3. Missing values
print("\n3. MISSING VALUES")
print("-" * 70)

missing = df.isnull().sum()

missing_table = pd.DataFrame({
    "Column": missing.index,
    "Missing_Values": missing.values
})

missing_table = missing_table[
    missing_table["Missing_Values"] > 0
]

if missing_table.empty:
    print("✓ No missing values found.")
else:
    display(missing_table)


# 4. Duplicate rows
print("\n4. DUPLICATE ROWS")
print("-" * 70)

duplicates = df.duplicated().sum()

print(f"Duplicate rows: {duplicates:,}")


# 5. Unique values
print("\n5. UNIQUE VALUES")
print("-" * 70)

for col in [
    "State",
    "District",
    "Crop",
    "Season",
    "Irrigation_Method"
]:

    if col in df.columns:

        print(
            f"{col}: "
            f"{df[col].nunique()} unique values"
        )


# 6. Statistical summary
print("\n6. NUMERICAL SUMMARY")
print("-" * 70)

display(df.describe().T)
# ============================================================
# EDA PART 2 — PRODUCTION & YIELD ANALYSIS
# ============================================================

print("=" * 70)
print("EDA PART 2 — PRODUCTION & YIELD ANALYSIS")
print("=" * 70)


# ============================================================
# 1. PRODUCTION BY CROP
# ============================================================

production_by_crop = (
    df.groupby("Crop")["Production_Tonnes"]
      .sum()
      .sort_values(ascending=False)
)

print("\n1. TOTAL PRODUCTION BY CROP")
print("-" * 70)

display(
    production_by_crop
    .rename_axis("Crop")
    .reset_index(name="Total_Production_Tonnes")
)


# Chart
plt.figure(figsize=(10, 6))

production_by_crop.sort_values().plot(
    kind="barh"
)

plt.title("Total Production by Crop")
plt.xlabel("Production (Tonnes)")
plt.ylabel("Crop")
plt.tight_layout()
plt.show()


# ============================================================
# 2. PRODUCTION BY STATE
# ============================================================

production_by_state = (
    df.groupby("State")["Production_Tonnes"]
      .sum()
      .sort_values(ascending=False)
)

print("\n2. TOTAL PRODUCTION BY STATE")
print("-" * 70)

display(
    production_by_state
    .rename_axis("State")
    .reset_index(name="Total_Production_Tonnes")
)


plt.figure(figsize=(10, 6))

production_by_state.sort_values().plot(
    kind="barh"
)

plt.title("Total Production by State")
plt.xlabel("Production (Tonnes)")
plt.ylabel("State")
plt.tight_layout()
plt.show()


# ============================================================
# 3. PRODUCTION BY SEASON
# ============================================================

production_by_season = (
    df.groupby("Season")["Production_Tonnes"]
      .sum()
      .sort_values(ascending=False)
)

print("\n3. TOTAL PRODUCTION BY SEASON")
print("-" * 70)

display(
    production_by_season
    .rename_axis("Season")
    .reset_index(name="Total_Production_Tonnes")
)


plt.figure(figsize=(8, 5))

production_by_season.plot(
    kind="bar"
)

plt.title("Total Production by Season")
plt.xlabel("Season")
plt.ylabel("Production (Tonnes)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# ============================================================
# 4. AVERAGE YIELD BY CROP
# ============================================================

yield_by_crop = (
    df.groupby("Crop")["Yield_Tonnes_Ha"]
      .mean()
      .sort_values(ascending=False)
)

print("\n4. AVERAGE YIELD BY CROP")
print("-" * 70)

display(
    yield_by_crop
    .rename_axis("Crop")
    .reset_index(name="Average_Yield_Tonnes_Ha")
)


plt.figure(figsize=(10, 6))

yield_by_crop.sort_values().plot(
    kind="barh"
)

plt.title("Average Yield by Crop")
plt.xlabel("Average Yield (Tonnes/Ha)")
plt.ylabel("Crop")
plt.tight_layout()
plt.show()


# ============================================================
# 5. AVERAGE YIELD BY STATE
# ============================================================

yield_by_state = (
    df.groupby("State")["Yield_Tonnes_Ha"]
      .mean()
      .sort_values(ascending=False)
)

print("\n5. AVERAGE YIELD BY STATE")
print("-" * 70)

display(
    yield_by_state
    .rename_axis("State")
    .reset_index(name="Average_Yield_Tonnes_Ha")
)


plt.figure(figsize=(10, 6))

yield_by_state.sort_values().plot(
    kind="barh"
)

plt.title("Average Yield by State")
plt.xlabel("Average Yield (Tonnes/Ha)")
plt.ylabel("State")
plt.tight_layout()
plt.show()
# ============================================================
# EDA PART 3 — REVENUE & PROFIT ANALYSIS
# ============================================================

print("=" * 70)
print("EDA PART 3 — REVENUE & PROFIT ANALYSIS")
print("=" * 70)


# ============================================================
# 1. REVENUE BY CROP
# ============================================================

revenue_by_crop = (
    df.groupby("Crop")["Revenue_INR"]
      .sum()
      .sort_values(ascending=False)
)

print("\n1. TOTAL REVENUE BY CROP")
print("-" * 70)

display(
    revenue_by_crop
    .rename_axis("Crop")
    .reset_index(name="Total_Revenue_INR")
)


plt.figure(figsize=(10, 6))

revenue_by_crop.sort_values().plot(
    kind="barh"
)

plt.title("Total Revenue by Crop")
plt.xlabel("Revenue (INR)")
plt.ylabel("Crop")
plt.tight_layout()
plt.show()


# ============================================================
# 2. PROFIT BY CROP
# ============================================================

profit_by_crop = (
    df.groupby("Crop")["Profit_INR"]
      .sum()
      .sort_values(ascending=False)
)

print("\n2. TOTAL PROFIT BY CROP")
print("-" * 70)

display(
    profit_by_crop
    .rename_axis("Crop")
    .reset_index(name="Total_Profit_INR")
)


plt.figure(figsize=(10, 6))

profit_by_crop.sort_values().plot(
    kind="barh"
)

plt.title("Total Profit by Crop")
plt.xlabel("Profit (INR)")
plt.ylabel("Crop")
plt.tight_layout()
plt.show()


# ============================================================
# 3. PROFIT MARGIN BY CROP
# ============================================================

crop_financials = (
    df.groupby("Crop")
      .agg(
          Revenue_INR=("Revenue_INR", "sum"),
          Profit_INR=("Profit_INR", "sum")
      )
)

crop_financials["Profit_Margin_pct"] = (
    crop_financials["Profit_INR"]
    / crop_financials["Revenue_INR"]
) * 100

crop_financials = crop_financials.sort_values(
    "Profit_Margin_pct",
    ascending=False
)

print("\n3. PROFIT MARGIN BY CROP")
print("-" * 70)

display(crop_financials.reset_index())


plt.figure(figsize=(10, 6))

crop_financials["Profit_Margin_pct"].sort_values().plot(
    kind="barh"
)

plt.title("Profit Margin by Crop")
plt.xlabel("Profit Margin (%)")
plt.ylabel("Crop")
plt.tight_layout()
plt.show()


# ============================================================
# 4. REVENUE BY SEASON
# ============================================================

revenue_by_season = (
    df.groupby("Season")["Revenue_INR"]
      .sum()
      .sort_values(ascending=False)
)

print("\n4. TOTAL REVENUE BY SEASON")
print("-" * 70)

display(
    revenue_by_season
    .rename_axis("Season")
    .reset_index(name="Total_Revenue_INR")
)


plt.figure(figsize=(8, 5))

revenue_by_season.plot(
    kind="bar"
)

plt.title("Total Revenue by Season")
plt.xlabel("Season")
plt.ylabel("Revenue (INR)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# ============================================================
# 5. PROFIT BY SEASON
# ============================================================

profit_by_season = (
    df.groupby("Season")["Profit_INR"]
      .sum()
      .sort_values(ascending=False)
)

print("\n5. TOTAL PROFIT BY SEASON")
print("-" * 70)

display(
    profit_by_season
    .rename_axis("Season")
    .reset_index(name="Total_Profit_INR")
)


plt.figure(figsize=(8, 5))

profit_by_season.plot(
    kind="bar"
)

plt.title("Total Profit by Season")
plt.xlabel("Season")
plt.ylabel("Profit (INR)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
# ============================================================
# EDA PART 4 — WATER & IRRIGATION ANALYSIS
# ============================================================

print("=" * 70)
print("EDA PART 4 — WATER & IRRIGATION ANALYSIS")
print("=" * 70)


# ============================================================
# 1. WATER USE BY CROP
# ============================================================

water_by_crop = (
    df.groupby("Crop")["Water_Used_m3"]
      .sum()
      .sort_values(ascending=False)
)

print("\n1. TOTAL WATER USE BY CROP")
print("-" * 70)

display(
    water_by_crop
    .rename_axis("Crop")
    .reset_index(name="Total_Water_Used_m3")
)


plt.figure(figsize=(10, 6))

water_by_crop.sort_values().plot(
    kind="barh"
)

plt.title("Total Water Use by Crop")
plt.xlabel("Water Used (m³)")
plt.ylabel("Crop")
plt.tight_layout()
plt.show()


# ============================================================
# 2. WATER PRODUCTIVITY BY CROP
# ============================================================

water_productivity_crop = (
    df.groupby("Crop")
      .agg(
          Production_Tonnes=("Production_Tonnes", "sum"),
          Water_Used_m3=("Water_Used_m3", "sum")
      )
)

water_productivity_crop["Water_Productivity_kg_m3"] = (
    water_productivity_crop["Production_Tonnes"]
    / water_productivity_crop["Water_Used_m3"]
) * 1000

water_productivity_crop = (
    water_productivity_crop
    .sort_values(
        "Water_Productivity_kg_m3",
        ascending=False
    )
)

print("\n2. WATER PRODUCTIVITY BY CROP")
print("-" * 70)

display(
    water_productivity_crop.reset_index()
)


plt.figure(figsize=(10, 6))

water_productivity_crop[
    "Water_Productivity_kg_m3"
].sort_values().plot(
    kind="barh"
)

plt.title("Water Productivity by Crop")
plt.xlabel("Water Productivity (kg/m³)")
plt.ylabel("Crop")
plt.tight_layout()
plt.show()


# ============================================================
# 3. IRRIGATION METHOD — PRODUCTION
# ============================================================

production_by_irrigation = (
    df.groupby("Irrigation_Method")["Production_Tonnes"]
      .sum()
      .sort_values(ascending=False)
)

print("\n3. PRODUCTION BY IRRIGATION METHOD")
print("-" * 70)

display(
    production_by_irrigation
    .rename_axis("Irrigation_Method")
    .reset_index(name="Total_Production_Tonnes")
)


plt.figure(figsize=(10, 6))

production_by_irrigation.plot(
    kind="bar"
)

plt.title("Production by Irrigation Method")
plt.xlabel("Irrigation Method")
plt.ylabel("Production (Tonnes)")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()


# ============================================================
# 4. IRRIGATION METHOD — YIELD
# ============================================================

yield_by_irrigation = (
    df.groupby("Irrigation_Method")["Yield_Tonnes_Ha"]
      .mean()
      .sort_values(ascending=False)
)

print("\n4. AVERAGE YIELD BY IRRIGATION METHOD")
print("-" * 70)

display(
    yield_by_irrigation
    .rename_axis("Irrigation_Method")
    .reset_index(name="Average_Yield_Tonnes_Ha")
)


plt.figure(figsize=(10, 6))

yield_by_irrigation.sort_values().plot(
    kind="barh"
)

plt.title("Average Yield by Irrigation Method")
plt.xlabel("Average Yield (Tonnes/Ha)")
plt.ylabel("Irrigation Method")
plt.tight_layout()
plt.show()


# ============================================================
# 5. IRRIGATION METHOD — WATER PRODUCTIVITY
# ============================================================

irrigation_water = (
    df.groupby("Irrigation_Method")
      .agg(
          Production_Tonnes=("Production_Tonnes", "sum"),
          Water_Used_m3=("Water_Used_m3", "sum")
      )
)

irrigation_water["Water_Productivity_kg_m3"] = (
    irrigation_water["Production_Tonnes"]
    / irrigation_water["Water_Used_m3"]
) * 1000

irrigation_water = irrigation_water.sort_values(
    "Water_Productivity_kg_m3",
    ascending=False
)

print("\n5. WATER PRODUCTIVITY BY IRRIGATION METHOD")
print("-" * 70)

display(irrigation_water.reset_index())


plt.figure(figsize=(10, 6))

irrigation_water[
    "Water_Productivity_kg_m3"
].sort_values().plot(
    kind="barh"
)

plt.title("Water Productivity by Irrigation Method")
plt.xlabel("Water Productivity (kg/m³)")
plt.ylabel("Irrigation Method")
plt.tight_layout()
plt.show()
# ============================================================
# EDA PART 5 — STATE, DISTRICT & CROP PERFORMANCE
# ============================================================

print("=" * 70)
print("EDA PART 5 — STATE, DISTRICT & CROP PERFORMANCE")
print("=" * 70)


# ============================================================
# 1. STATE PERFORMANCE
# ============================================================

state_performance = (
    df.groupby("State")
      .agg(
          Production_Tonnes=("Production_Tonnes", "sum"),
          Average_Yield=("Yield_Tonnes_Ha", "mean"),
          Revenue_INR=("Revenue_INR", "sum"),
          Profit_INR=("Profit_INR", "sum"),
          Water_Used_m3=("Water_Used_m3", "sum")
      )
)

state_performance["Profit_Margin_pct"] = (
    state_performance["Profit_INR"]
    / state_performance["Revenue_INR"]
) * 100

state_performance["Water_Productivity_kg_m3"] = (
    state_performance["Production_Tonnes"]
    / state_performance["Water_Used_m3"]
) * 1000


print("\n1. STATE PERFORMANCE")
print("-" * 70)

display(
    state_performance
    .sort_values(
        "Production_Tonnes",
        ascending=False
    )
    .reset_index()
)


# ============================================================
# 2. TOP 5 STATES BY PRODUCTION
# ============================================================

top_states = (
    state_performance
    .nlargest(5, "Production_Tonnes")
)

print("\n2. TOP 5 STATES BY PRODUCTION")
print("-" * 70)

display(
    top_states[
        [
            "Production_Tonnes",
            "Average_Yield",
            "Revenue_INR",
            "Profit_INR"
        ]
    ].reset_index()
)


# ============================================================
# 3. DISTRICT PERFORMANCE
# ============================================================

district_performance = (
    df.groupby("District")
      .agg(
          Production_Tonnes=("Production_Tonnes", "sum"),
          Average_Yield=("Yield_Tonnes_Ha", "mean"),
          Revenue_INR=("Revenue_INR", "sum"),
          Profit_INR=("Profit_INR", "sum"),
          Water_Used_m3=("Water_Used_m3", "sum")
      )
)

district_performance["Profit_Margin_pct"] = (
    district_performance["Profit_INR"]
    / district_performance["Revenue_INR"]
) * 100


print("\n3. TOP 10 DISTRICTS BY PRODUCTION")
print("-" * 70)

display(
    district_performance
    .nlargest(10, "Production_Tonnes")
    .reset_index()
)


# ============================================================
# 4. CROP PERFORMANCE SCORECARD
# ============================================================

crop_performance = (
    df.groupby("Crop")
      .agg(
          Production_Tonnes=("Production_Tonnes", "sum"),
          Average_Yield=("Yield_Tonnes_Ha", "mean"),
          Revenue_INR=("Revenue_INR", "sum"),
          Profit_INR=("Profit_INR", "sum"),
          Water_Used_m3=("Water_Used_m3", "sum")
      )
)

crop_performance["Profit_Margin_pct"] = (
    crop_performance["Profit_INR"]
    / crop_performance["Revenue_INR"]
) * 100

crop_performance["Water_Productivity_kg_m3"] = (
    crop_performance["Production_Tonnes"]
    / crop_performance["Water_Used_m3"]
) * 1000


print("\n4. CROP PERFORMANCE SCORECARD")
print("-" * 70)

display(
    crop_performance
    .sort_values(
        "Production_Tonnes",
        ascending=False
    )
    .reset_index()
)
# ============================================================
# EDA PART 6 — CORRELATION & RELATIONSHIP ANALYSIS
# ============================================================

print("=" * 70)
print("EDA PART 6 — CORRELATION ANALYSIS")
print("=" * 70)


```

```python



# ==========================================
# 1. DEFINE CORRECT STATE-DISTRICT MAPPING
# ==========================================

state_district_mapping = {
    "Andhra Pradesh": ["Guntur", "Krishna"],
    "Telangana": ["Warangal", "Nalgonda"],
    "Maharashtra": ["Nashik"],
    "Madhya Pradesh": ["Indore"],
    "Karnataka": ["Raichur"],
    "Gujarat": ["Rajkot"],
    "Tamil Nadu": ["Erode"],
    "Punjab": ["Ludhiana"]
}


# ==========================================
# 2. CORRECT DISTRICTS BASED ON STATE
# ==========================================

# Fixed random seed so the result is reproducible
rng = np.random.default_rng(42)

for state, districts in state_district_mapping.items():

    # Select rows belonging to this state
    mask = df["State"] == state

    # If state has only one valid district
    if len(districts) == 1:

        df.loc[mask, "District"] = districts[0]

    # If state has multiple valid districts
    else:

        df.loc[mask, "District"] = rng.choice(
            districts,
            size=mask.sum(),
            replace=True
        )


# ==========================================
# 3. VALIDATE STATE-DISTRICT MATCHING
# ==========================================

print("\nChecking State-District Matching...\n")

for state, districts in state_district_mapping.items():

    state_data = df[df["State"] == state]

    invalid_districts = state_data[
        ~state_data["District"].isin(districts)
    ]["District"].unique()

    if len(invalid_districts) == 0:
        print(f"✓ {state}: Correct")
    else:
        print(f"✗ {state}: Invalid districts -> {invalid_districts}")


# ==========================================
# 4. DISPLAY STATE-DISTRICT CROSS TAB
# ==========================================

print("\nState-District Distribution:\n")

state_district_table = pd.crosstab(
    df["State"],
    df["District"]
)

display(state_district_table)


# ==========================================
# 5. FINAL VALIDATION
# ==========================================

all_valid = True

for state, districts in state_district_mapping.items():

    mask = df["State"] == state

    if not df.loc[mask, "District"].isin(districts).all():
        all_valid = False


if all_valid:
    print("\n✓ ALL STATE-DISTRICT COMBINATIONS ARE VALID")
else:
    print("\n✗ SOME INVALID COMBINATIONS STILL EXIST")


# ==========================================
# 6. SAVE CORRECTED DATASET
# ==========================================

output_file = "final_dataset_state_district_corrected.csv"

df.to_csv(output_file, index=False)

print("\nCorrected dataset saved successfully as:")
print(output_file)

print("\nFinal Dataset Shape:", df.shape)

# Load corrected dataset

# 1. State distribution
print("\nState Distribution:")
print(df["State"].value_counts())

# 2. District distribution
print("\nDistrict Distribution:")
print(df["District"].value_counts())

# 3. Season distribution
print("\nSeason Distribution:")
print(df["Season"].value_counts())

# 4. Crop distribution
print("\nCrop Distribution:")
print(df["Crop"].value_counts())
```

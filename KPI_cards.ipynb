```python
# ============================================================
# PROFESSIONAL KPI CALCULATION
# ============================================================

import pandas as pd
import numpy as np


# ------------------------------------------------------------
# 1. LOAD CLEANED DATASET
# ------------------------------------------------------------

input_path = "seasonal_agriculture_performance_dataset_cleaned.csv"

df = pd.read_csv(input_path)

print("Original Dataset Shape:", df.shape)


# ------------------------------------------------------------
# 2. REQUIRED SOURCE COLUMNS
# ------------------------------------------------------------

production_col = "Production_Tonnes"
yield_col = "Yield_Tonnes_Ha"
revenue_col = "Revenue_INR"
profit_col = "Profit_INR"
water_col = "Water_Used_m3"

required_columns = [
    production_col,
    yield_col,
    revenue_col,
    profit_col,
    water_col
]


# ------------------------------------------------------------
# 3. CHECK REQUIRED COLUMNS
# ------------------------------------------------------------

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Required columns are missing: {missing_columns}"
    )

print("✓ All required columns are available.")


# ------------------------------------------------------------
# 4. CONVERT KPI SOURCE COLUMNS TO NUMERIC
# ------------------------------------------------------------

for col in required_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")


# ------------------------------------------------------------
# 5. CHECK MISSING VALUES
# ------------------------------------------------------------

missing_values = df[required_columns].isna().sum()

print("\nMissing Values in KPI Columns:")
print(missing_values)

if missing_values.sum() > 0:
    raise ValueError(
        "KPI calculation stopped because missing values exist."
    )


# ------------------------------------------------------------
# 6. CHECK INFINITE VALUES
# ------------------------------------------------------------

if np.isinf(df[required_columns].to_numpy()).any():
    raise ValueError(
        "KPI calculation stopped because infinite values exist."
    )

print("✓ No missing or infinite values found.")


# ============================================================
# KPI CALCULATIONS
# ============================================================


# ------------------------------------------------------------
# KPI 1 — TOTAL PRODUCTION
# ------------------------------------------------------------

total_production = df[production_col].sum()


# ------------------------------------------------------------
# KPI 2 — AVERAGE YIELD
# ------------------------------------------------------------

average_yield = df[yield_col].mean()


# ------------------------------------------------------------
# KPI 3 — TOTAL REVENUE
# ------------------------------------------------------------

total_revenue = df[revenue_col].sum()

# Convert INR to Billion INR
total_revenue_billion = total_revenue / 1_000_000_000


# ------------------------------------------------------------
# KPI 4 — PROFIT MARGIN
# ------------------------------------------------------------

total_profit = df[profit_col].sum()

if total_revenue == 0:
    raise ZeroDivisionError(
        "Total Revenue is zero. Profit Margin cannot be calculated."
    )

profit_margin = (
    total_profit / total_revenue
) * 100


# ------------------------------------------------------------
# KPI 5 — WATER PRODUCTIVITY
# ------------------------------------------------------------

total_water = df[water_col].sum()

if total_water <= 0:
    raise ValueError(
        "Total Water Used must be greater than zero."
    )

# Tonnes per cubic metre
water_productivity_tonnes_m3 = (
    total_production / total_water
)

# Convert tonnes to kilograms
water_productivity_kg_m3 = (
    water_productivity_tonnes_m3 * 1000
)


# ============================================================
# PROFESSIONAL KPI CARDS
# ============================================================

print("\n" + "=" * 60)
print("PROFESSIONAL KPI CARDS")
print("=" * 60)


print("\n1. Total Production")
print(f"   {total_production:,.2f} Tonnes")


print("\n2. Average Yield")
print(f"   {average_yield:,.2f} Tonnes/Ha")


print("\n3. Total Revenue")
print(f"   ₹{total_revenue_billion:,.2f} Billion")


print("\n4. Profit Margin")
print(f"   {profit_margin:,.2f}%")


print("\n5. Water Productivity")
print(f"   {water_productivity_kg_m3:,.3f} kg/m³")


# ============================================================
# FINAL VALIDATION
# ============================================================

print("\n" + "=" * 60)
print("FINAL KPI VALIDATION")
print("=" * 60)


assert np.isfinite(total_production)
assert np.isfinite(average_yield)
assert np.isfinite(total_revenue)
assert np.isfinite(profit_margin)
assert np.isfinite(water_productivity_kg_m3)

assert total_production >= 0
assert total_revenue >= 0
assert total_water > 0


print("✓ Total Production calculation verified")
print("✓ Average Yield calculation verified")
print("✓ Total Revenue calculation verified")
print("✓ Profit Margin calculation verified")
print("✓ Water Productivity calculation verified")

print("\nALL 5 KPI CALCULATIONS PASSED SUCCESSFULLY.")


# ============================================================
# ADD KPI VALUES TO CLEANED DATASET
# ============================================================

df["Total Production"] = total_production

df["Average Yield"] = average_yield

df["Total Revenue"] = total_revenue_billion

df["Profit Margin"] = profit_margin

df["Water Productivity"] = water_productivity_kg_m3


# ============================================================
# DISPLAY FINAL KPI COLUMNS
# ============================================================

print("\n" + "=" * 60)
print("KPI COLUMNS ADDED TO DATASET")
print("=" * 60)

print([
    "Total Production",
    "Average Yield",
    "Total Revenue",
    "Profit Margin",
    "Water Productivity"
])


# ============================================================
# SAVE AS A NEW FILE
# ============================================================

output_path = (
    "seasonal_agriculture_performance_dataset_cleaned_with_kpi.csv"
)

df.to_csv(output_path, index=False)


# ============================================================
# FINAL CHECK
# ============================================================

print("\n" + "=" * 60)
print("FINAL DATASET")
print("=" * 60)

print("Final Dataset Shape:", df.shape)

print("\n✓ Original dataset was not overwritten")
print("✓ Existing rows preserved")
print("✓ Existing columns preserved")
print("✓ 5 KPI columns added at the end")
print(f"✓ File saved as: {output_path}")


# Show KPI values
print("\nFINAL KPI VALUES:")

display(
    df[
        [
            "Total Production",
            "Average Yield",
            "Total Revenue",
            "Profit Margin",
            "Water Productivity"
        ]
    ].head()
)
```

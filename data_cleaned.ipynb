```python
# ============================================================
# 1. FILE NAMES
# ============================================================

INPUT_FILE = "seasonal_agriculture_performance_dataset.csv"
OUTPUT_FILE = "seasonal_agriculture_performance_dataset_cleaned.csv"
```

```python
# ============================================================
# 2. COLUMN DEFINITIONS
# ============================================================

TEXT_COLUMNS = [
    "Farm_ID",
    "State",
    "District",
    "Crop",
    "Season",
    "Irrigation_Method"
]

NUMERIC_COLUMNS = [
    "Farm_Area_Hectares",
    "Rainfall_mm",
    "Avg_Temperature_C",
    "Humidity_pct",
    "Sunlight_Hours_Day",
    "Soil_pH",
    "Soil_Moisture_pct",
    "Nitrogen_kg_ha",
    "Phosphorus_kg_ha",
    "Potassium_kg_ha",
    "Fertilizer_kg_ha",
    "Pesticide_Litre_ha",
    "Seed_Quality_Score",
    "Yield_Tonnes_Ha",
    "Production_Tonnes",
    "Market_Price_INR_Tonne",
    "Total_Cost_INR",
    "Revenue_INR",
    "Profit_INR",
    "Water_Used_m3",
    "Water_Efficiency_t_per_1000m3",
    "Disease_Pest_Risk_pct"
]

EXPECTED_COLUMNS = [
    "Farm_ID",
    "State",
    "District",
    "Crop",
    "Season",
    "Farm_Area_Hectares",
    "Rainfall_mm",
    "Avg_Temperature_C",
    "Humidity_pct",
    "Sunlight_Hours_Day",
    "Soil_pH",
    "Soil_Moisture_pct",
    "Nitrogen_kg_ha",
    "Phosphorus_kg_ha",
    "Potassium_kg_ha",
    "Irrigation_Method",
    "Fertilizer_kg_ha",
    "Pesticide_Litre_ha",
    "Seed_Quality_Score",
    "Yield_Tonnes_Ha",
    "Production_Tonnes",
    "Market_Price_INR_Tonne",
    "Total_Cost_INR",
    "Revenue_INR",
    "Profit_INR",
    "Water_Used_m3",
    "Water_Efficiency_t_per_1000m3",
    "Disease_Pest_Risk_pct"
]
```

```python
# ============================================================
# 3. LOAD RAW DATASET
# ============================================================

print("=" * 70)
print("SEASONAL AGRICULTURE DATASET CLEANING")
print("=" * 70)

df = pd.read_csv(INPUT_FILE)

print("\nRAW DATASET INFORMATION")
print("-" * 70)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python

# ============================================================
# 4. CHECK COLUMN STRUCTURE
# ============================================================
if df.columns.tolist() != EXPECTED_COLUMNS:

    print("\nERROR: Column structure is different from expected.")

    missing_columns = [
        column
        for column in EXPECTED_COLUMNS
        if column not in df.columns
    ]

    extra_columns = [
        column
        for column in df.columns
        if column not in EXPECTED_COLUMNS
    ]

    if missing_columns:
        print("Missing columns:", missing_columns)

    if extra_columns:
        print("Extra columns:", extra_columns)

    raise ValueError(
        "Dataset columns do not match the expected structure."
    )

print("Column structure: OK")
```

```python

# ============================================================
# 5. CHECK ORIGINAL DATA QUALITY
# ============================================================

original_rows = len(df)

original_missing = int(
    df.isna().sum().sum()
)

original_duplicates = int(
    df.duplicated().sum()
)

original_farm_duplicates = int(
    df["Farm_ID"].duplicated().sum()
)

print("\nORIGINAL DATA QUALITY")
print("-" * 70)
print("Missing values:", original_missing)
print("Duplicate rows:", original_duplicates)
print("Duplicate Farm_ID:", original_farm_duplicates)
```

```python
#============================================================
# 6. STANDARDIZE TEXT COLUMNS
# ============================================================

print("\nSTANDARDIZING TEXT COLUMNS...")

for column in TEXT_COLUMNS:

    df[column] = df[column].astype("string")

    # Remove leading and trailing spaces
    df[column] = df[column].str.strip()

    # Replace multiple spaces with one space
    df[column] = df[column].str.replace(
        r"\s+",
        " ",
        regex=True
    )

    # Check for missing or blank values
    blank_values = (
        df[column].isna()
        |
        (df[column].str.len() == 0)
    )

    if blank_values.any():
        raise ValueError(
            f"Blank or missing value found in {column}."
        )

print("Text standardization: COMPLETED")
```

```python
# ============================================================
# 7. CONVERT NUMERIC COLUMNS
# ============================================================

print("\nCONVERTING NUMERIC COLUMNS...")

for column in NUMERIC_COLUMNS:

    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )

print("Numeric conversion: COMPLETED")


# ============================================================
# 8. HANDLE MISSING YIELD VALUES
# ============================================================

print("\nCHECKING MISSING YIELD VALUES...")

yield_missing = df["Yield_Tonnes_Ha"].isna()

if yield_missing.any():

    valid_yield = (
        df.loc[
            yield_missing,
            "Production_Tonnes"
        ].notna()
        &
        df.loc[
            yield_missing,
            "Farm_Area_Hectares"
        ].notna()
        &
        (
            df.loc[
                yield_missing,
                "Farm_Area_Hectares"
            ] > 0
        )
    )

    if not valid_yield.all():
        raise ValueError(
            "Some missing Yield values cannot be calculated safely."
        )

    df.loc[
        yield_missing,
        "Yield_Tonnes_Ha"
    ] = (
        df.loc[
            yield_missing,
            "Production_Tonnes"
        ]
        /
        df.loc[
            yield_missing,
            "Farm_Area_Hectares"
        ]
    ).round(2)

    print("Missing Yield values calculated.")

else:

    print("No missing Yield values found.")


# ============================================================
# 9. HANDLE MISSING ENVIRONMENTAL VALUES
# ============================================================

print("\nHANDLING ENVIRONMENTAL MISSING VALUES...")

for column in [
    "Rainfall_mm",
    "Soil_Moisture_pct"
]:

    missing_before = int(
        df[column].isna().sum()
    )

    if missing_before > 0:

        # First level:
        # State + Crop + Season median

        df[column] = df[column].fillna(
            df.groupby(
                ["State", "Crop", "Season"]
            )[column].transform("median")
        )

        # Second level:
        # Crop + Season median

        df[column] = df[column].fillna(
            df.groupby(
                ["Crop", "Season"]
            )[column].transform("median")
        )

        # Third level:
        # Overall median

        df[column] = df[column].fillna(
            df[column].median()
        )

        missing_after = int(
            df[column].isna().sum()
        )

        print(
            f"{column}: "
            f"{missing_before} missing -> "
            f"{missing_after} remaining"
        )

    else:

        print(
            f"{column}: No missing values"
        )


# ============================================================
# 10. REMOVE DUPLICATE ROWS
# ============================================================

print("\nCHECKING DUPLICATE ROWS...")

duplicate_count = int(
    df.duplicated().sum()
)

if duplicate_count > 0:

    print(
        f"Found {duplicate_count} duplicate rows."
    )

    df = df.drop_duplicates(
        keep="first"
    ).reset_index(drop=True)

    print("Duplicate rows removed.")

else:

    print("No duplicate rows found.")


# ============================================================
# 11. CHECK DUPLICATE FARM IDs
# ============================================================

farm_id_duplicates = int(
    df["Farm_ID"].duplicated().sum()
)

if farm_id_duplicates > 0:

    raise ValueError(
        "Duplicate Farm_ID values remain. "
        "Please inspect the dataset."
    )

print("Farm_ID uniqueness: OK")


# ============================================================
# 12. CHECK REMAINING MISSING VALUES
# ============================================================

remaining_missing = int(
    df[NUMERIC_COLUMNS]
    .isna()
    .sum()
    .sum()
)

if remaining_missing > 0:

    missing_details = (
        df[NUMERIC_COLUMNS]
        .isna()
        .sum()
    )

    missing_details = missing_details[
        missing_details > 0
    ]

    print(
        "\nRemaining missing numeric values:"
    )

    print(missing_details)

    raise ValueError(
        "Missing numeric values remain."
    )

print("\nMissing numeric values: 0")


# ============================================================
# 13. CHECK NON-FINITE VALUES
# ============================================================

numeric_array = df[
    NUMERIC_COLUMNS
].to_numpy(dtype=float)

if not np.isfinite(
    numeric_array
).all():

    raise ValueError(
        "Non-finite numeric values such as "
        "inf or -inf were found."
    )

print("Non-finite numeric values: 0")


# ============================================================
# 14. VALIDATE NUMERIC RANGES
# ============================================================

print("\nVALIDATING NUMERIC RANGES...")

range_checks = {

    "Farm_Area_Hectares":
        df["Farm_Area_Hectares"] > 0,

    "Rainfall_mm":
        df["Rainfall_mm"] >= 0,

    "Humidity_pct":
        df["Humidity_pct"].between(0, 100),

    "Sunlight_Hours_Day":
        df["Sunlight_Hours_Day"].between(0, 24),

    "Soil_pH":
        df["Soil_pH"].between(0, 14),

    "Soil_Moisture_pct":
        df["Soil_Moisture_pct"].between(0, 100),

    "Seed_Quality_Score":
        df["Seed_Quality_Score"].between(0, 1),

    "Disease_Pest_Risk_pct":
        df["Disease_Pest_Risk_pct"].between(0, 100),

    "Nitrogen_kg_ha":
        df["Nitrogen_kg_ha"] >= 0,

    "Phosphorus_kg_ha":
        df["Phosphorus_kg_ha"] >= 0,

    "Potassium_kg_ha":
        df["Potassium_kg_ha"] >= 0,

    "Fertilizer_kg_ha":
        df["Fertilizer_kg_ha"] >= 0,

    "Pesticide_Litre_ha":
        df["Pesticide_Litre_ha"] >= 0,

    "Production_Tonnes":
        df["Production_Tonnes"] >= 0,

    "Market_Price_INR_Tonne":
        df["Market_Price_INR_Tonne"] >= 0,

    "Total_Cost_INR":
        df["Total_Cost_INR"] >= 0,

    "Revenue_INR":
        df["Revenue_INR"] >= 0,

    "Water_Used_m3":
        df["Water_Used_m3"] > 0
}


for column, condition in range_checks.items():

    if not condition.all():

        invalid_count = int(
            (~condition).sum()
        )

        raise ValueError(
            f"{column} contains "
            f"{invalid_count} invalid values."
        )

print("Numeric range validation: PASSED")


# ============================================================
# 15. VALIDATE PRODUCTION
# ============================================================

production_difference = (
    df["Production_Tonnes"]
    -
    (
        df["Farm_Area_Hectares"]
        *
        df["Yield_Tonnes_Ha"]
    )
).abs()

if not (
    production_difference <= 0.005001
).all():

    raise ValueError(
        "Production values do not match "
        "Farm Area × Yield."
    )

print("Production calculation: PASSED")


# ============================================================
# 16. VALIDATE REVENUE
# ============================================================

revenue_difference = (
    df["Revenue_INR"]
    -
    (
        df["Production_Tonnes"]
        *
        df["Market_Price_INR_Tonne"]
    )
).abs()

if not (
    revenue_difference <= 0.500001
).all():

    raise ValueError(
        "Revenue values do not match "
        "Production × Market Price."
    )

print("Revenue calculation: PASSED")


# ============================================================
# 17. VALIDATE PROFIT
# ============================================================

profit_difference = (
    df["Profit_INR"]
    -
    (
        df["Revenue_INR"]
        -
        df["Total_Cost_INR"]
    )
).abs()

if not (
    profit_difference <= 0.01
).all():

    raise ValueError(
        "Profit values do not match "
        "Revenue - Total Cost."
    )

print("Profit calculation: PASSED")


# ============================================================
# 18. VALIDATE WATER EFFICIENCY
# ============================================================

water_efficiency_difference = (
    df["Water_Efficiency_t_per_1000m3"]
    -
    (
        df["Production_Tonnes"]
        /
        df["Water_Used_m3"]
        *
        1000
    )
).abs()

if not (
    water_efficiency_difference <= 0.000501
).all():

    raise ValueError(
        "Water efficiency values are inconsistent."
    )

print("Water efficiency calculation: PASSED")


# ============================================================
# 19. FORMAT FINANCIAL COLUMNS
# ============================================================

print("\nFORMATTING FINANCIAL COLUMNS...")

financial_columns = [
    "Market_Price_INR_Tonne",
    "Total_Cost_INR",
    "Revenue_INR",
    "Profit_INR",
    "Water_Used_m3"
]

for column in financial_columns:

    df[column] = (
        df[column]
        .round()
        .astype("int64")
    )

print("Financial columns formatted.")


# ============================================================
# 20. RESTORE EXPECTED COLUMN ORDER
# ============================================================

df = df[
    EXPECTED_COLUMNS
]


# ============================================================
# 21. FINAL VALIDATION BEFORE EXPORT
# ============================================================

print("\nFINAL VALIDATION BEFORE EXPORT")
print("-" * 70)

assert df.columns.tolist() == EXPECTED_COLUMNS

assert not df.isna().any().any()

assert not df.duplicated().any()

assert not df["Farm_ID"].duplicated().any()

assert np.isfinite(
    df[NUMERIC_COLUMNS]
    .to_numpy(dtype=float)
).all()

print("Column structure: PASSED")
print("Missing values: PASSED")
print("Duplicate rows: PASSED")
print("Duplicate Farm_ID: PASSED")
print("Numeric values: PASSED")


# ============================================================
# 22. SAVE CLEANED CSV
# ============================================================

df.to_csv(
    OUTPUT_FILE,
    index=False,
    encoding="utf-8",
    lineterminator="\n"
)

print("\nCleaned CSV saved successfully.")


# ============================================================
# 23. READ THE SAVED CSV AGAIN
# ============================================================

print("\nVERIFYING SAVED CSV")
print("-" * 70)

cleaned = pd.read_csv(
    OUTPUT_FILE
)


# ============================================================
# 24. VERIFY EXPORTED DATASET
# ============================================================

assert cleaned.shape == df.shape

assert cleaned.columns.tolist() == EXPECTED_COLUMNS

assert not cleaned.isna().any().any()

assert not cleaned.duplicated().any()

assert not cleaned["Farm_ID"].duplicated().any()

assert np.isfinite(
    cleaned[NUMERIC_COLUMNS]
    .to_numpy(dtype=float)
).all()


# ============================================================
# 25. FINAL REPORT
# ============================================================

final_missing = int(
    cleaned.isna().sum().sum()
)

final_duplicates = int(
    cleaned.duplicated().sum()
)

final_farm_duplicates = int(
    cleaned["Farm_ID"].duplicated().sum()
)

print("\n" + "=" * 70)
print("DATASET CLEANING COMPLETED SUCCESSFULLY")
print("=" * 70)

print("Original rows     :", original_rows)
print("Final rows        :", len(cleaned))
print("Final columns     :", len(cleaned.columns))
print("Missing values    :", final_missing)
print("Duplicate rows    :", final_duplicates)
print("Duplicate Farm_ID :", final_farm_duplicates)

print("\nINPUT FILE:")
print(INPUT_FILE)

print("\nOUTPUT FILE:")
print(OUTPUT_FILE)

print("\nFINAL STATUS: CLEANED CSV VERIFIED SUCCESSFULLY")
print("=" * 70)
```
